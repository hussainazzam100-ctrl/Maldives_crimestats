```python
import pandas as pd 

file_path = 'MALDIVES_CRIMESTATS_2426.xlsx'
df = pd.read_excel(file_path, skiprows=2)

print(df)
```

ANALYSIS OF TRENDS AND ACROSS MULTPILE CRIMES OVER THE YEARS

```python
import matplotlib.pyplot as plt

row = df.iloc[0]
trend_data = pd.to_numeric(row.iloc[1:4])
x_labels = ['2024', '2025', '2026']

bar_colors = ['#FF6B6B', '#4D96FF', '#6BCB77']

fig, ax = plt.subplots(figsize=(7, 5))

bars = ax.bar(
    x_labels, trend_data.values,
    color=bar_colors, width=0.4, edgecolor='black'
)

ax.bar_label(bars, padding=2, fontweight='bold', fmt='%.0f')

ax.set_title('Theft Crime Stats (2024-2026)', fontsize=14, fontweight='bold')
ax.set_xlabel('Year', fontsize=11, labelpad=10)
ax.set_ylabel('Reported Crime Count', fontsize=11, labelpad=10)
ax.grid(axis='y', linestyle='--', alpha=0.5)
ax.set_axisbelow(True)


plt.show()
```
